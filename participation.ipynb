# Participation Analysis

```python
import pandas as pd
import os
from matplotlib import pyplot as plt

dataPath = os.path.join(os.getcwd(), 'Clean Data', 'completeResults.csv')

df = pd.read_csv(dataPath)
df.columns
```

## Attendance

```python
attendance = df.pivot_table(index='Leg', columns='Country', values='Score', aggfunc='count')
attendance = attendance.fillna(0)

figAtt, axAtt = plt.subplots(figsize=(10, 6))
attendance.plot(kind='bar', stacked=True, ax=axAtt)
```

## Medals

```python
placings = df['Category Rank'].value_counts().sort_index()
placings.plot(kind='bar', figsize=(10, 6))
```
